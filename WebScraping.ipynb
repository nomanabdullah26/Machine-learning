```python
webpage=requests.get('https://neocities.org/browse?page=1').text
```

```python
webpage
```

```python
soup =BeautifulSoup(webpage,'lxml')
```

```python
print(soup.prettify())
```

# **Way-1**

```python
site_names = [a.text.strip() for a in soup.select('.title a')]
```

```python
site_names
```

```python
print("Neocities sites Name :")
for i, name in enumerate(site_names, 1):
    print(f"{i}. {name}")
```
