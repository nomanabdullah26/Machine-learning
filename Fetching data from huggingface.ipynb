```python
# ============================================================
# ১ কোটি ডেটা নিয়ে TGNN মডেল ট্রেইন করার সম্পূর্ণ কোড
# ============================================================

# 1. প্রয়োজনীয় লাইব্রেরি ইনস্টল করা
# ============================================================
#!pip install torch torch-geometric torch-scatter torch-sparse torch-cluster torch-spline-conv -f https://data.pyg.org/whl/torch-2.0.0+cu118.html
#!pip install numpy pandas matplotlib scikit-learn networkx imbalanced-learn tqdm

# 2. লাইব্রেরি ইমপোর্ট
# ============================================================
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import GCNConv, global_mean_pool
from torch_geometric.data import Data, DataLoader
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, f1_score
from imblearn.under_sampling import RandomUnderSampler
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')

print("✅ সব লাইব্রেরি ইমপোর্ট হয়েছে!")

# 3. ডেটাসেট লোড করা (১ কোটি ডেটা)
# ============================================================
from datasets import load_dataset

print("\n🔄 ডেটাসেট লোড হচ্ছে...")
dataset = load_dataset("lacg030175/CIC-IoT-2023-neto-full", "random_3way")

# ১ কোটি ডেটা নিন (প্রথম 10,000,000 সারি)
TOTAL_SAMPLES = 10000000
print(f"\n📊 মোট {TOTAL_SAMPLES:,} ডেটা নেওয়া হচ্ছে...")

train_data = dataset['train'].select(range(TOTAL_SAMPLES))
print(f"✅ ট্রেইন ডেটা লোড হয়েছে: {len(train_data):,} সারি")

# 4. ডেটাকে Pandas DataFrame-এ কনভার্ট করা
# ============================================================
print("\n🔄 ডেটাকে Pandas DataFrame-এ কনভার্ট করা হচ্ছে...")
df = train_data.to_pandas()
print(f"✅ DataFrame তৈরি হয়েছে: {df.shape[0]:,} সারি × {df.shape[1]} কলাম")

# 5. ফিচার নির্বাচন করা
# ============================================================
print("\n📋 ফিচার নির্বাচন করা হচ্ছে...")

# প্রয়োজনীয় ফিচার (শীর্ষ ১৫টি গুরুত্বপূর্ণ ফিচার)
feature_cols = [
    'flow_duration', 'Protocol Type', 'Duration', 'Rate',
    'Srate', 'Drate', 'TCP', 'UDP', 'ICMP',
    'IAT', 'Tot sum', 'Min', 'Max', 'AVG', 'Std'
]

# টার্গেট লেবেল
target_col = 'label'  # 0=Benign, 1=Attack

# NaN মানগুলো 0 দিয়ে ফিল করুন
df[feature_cols] = df[feature_cols].fillna(0)

X = df[feature_cols].values
y = df[target_col].values

print(f"✅ ফিচার ম্যাট্রিক্স: {X.shape}")
print(f"✅ লেবেল ভেক্টর: {y.shape}")

# 6. ডেটা স্কেলিং করা
# ============================================================
print("\n📊 ডেটা স্কেলিং করা হচ্ছে...")
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(f"✅ স্কেলিং সম্পূর্ণ: {X_scaled.shape}")

# 7. ইমব্যালেন্সড ডেটা হ্যান্ডলিং (আন্ডার-স্যাম্পলিং)
# ============================================================
print("\n⚖️ ইমব্যালেন্সড ডেটা ব্যালেন্স করা হচ্ছে...")

# ক্লাস বণ্টন দেখুন
benign_count = sum(y == 0)
attack_count = sum(y == 1)
print(f"Before balancing - Benign: {benign_count:,}, Attack: {attack_count:,}")

# আন্ডার-স্যাম্পলিং
if benign_count > 0 and attack_count > 0:
    undersampler = RandomUnderSampler(sampling_strategy='auto', random_state=42)
    X_resampled, y_resampled = undersampler.fit_resample(X_scaled, y)

    # ব্যালেন্সড ডেটার সাইজ দেখুন
    benign_balanced = sum(y_resampled == 0)
    attack_balanced = sum(y_resampled == 1)
    print(f"After balancing - Benign: {benign_balanced:,}, Attack: {attack_balanced:,}")
    print(f"মোট ব্যালেন্সড ডেটা: {len(y_resampled):,}")
else:
    # যদি কোনো ক্লাস না থাকে
    X_resampled, y_resampled = X_scaled, y
    print("⚠️ কোনো ক্লাস খালি নেই, ব্যালেন্সিং করা হয়নি")

# 8. ট্রেইন-টেস্ট স্প্লিট
# ============================================================
print("\n🔪 ট্রেইন-টেস্ট স্প্লিট করা হচ্ছে...")
X_train, X_test, y_train, y_test = train_test_split(
    X_resampled, y_resampled, test_size=0.2, random_state=42, stratify=y_resampled
)
print(f"Train: {X_train.shape[0]:,}, Test: {X_test.shape[0]:,}")

# 9. টেম্পোরাল গ্রাফ তৈরি করা
# ============================================================
print("\n🕸️ টেম্পোরাল গ্রাফ তৈরি করা হচ্ছে...")

def create_temporal_graph(data, labels, time_steps=20):
    """
    ডেটা থেকে টেম্পোরাল গ্রাফ তৈরি করুন
    """
    graphs = []
    n_samples = len(data)
    samples_per_step = n_samples // time_steps

    for t in range(time_steps):
        start_idx = t * samples_per_step
        end_idx = (t + 1) * samples_per_step if t < time_steps - 1 else n_samples

        # এই টাইমস্টেপের ডেটা
        step_data = data[start_idx:end_idx]
        step_labels = labels[start_idx:end_idx]

        if len(step_data) == 0:
            continue

        # নোড ফিচার
        node_features = torch.tensor(step_data, dtype=torch.float)

        # এজ তৈরি করুন (কোসাইন সিমিলারিটি ব্যবহার করে)
        from sklearn.metrics.pairwise import cosine_similarity
        similarity = cosine_similarity(step_data)
        # ০.৭ এর বেশি সিমিলার থাকলে এজ তৈরি করুন
        edge_indices = np.where(similarity > 0.7)
        edge_index = torch.tensor(edge_indices, dtype=torch.long)

        # যদি কোনো এজ না থাকে, তাহলে ডামি এজ যোগ করুন
        if edge_index.size(1) == 0:
            edge_index = torch.tensor([[0], [0]], dtype=torch.long)

        # লেবেল
        labels_tensor = torch.tensor(step_labels, dtype=torch.long)

        # PyTorch Geometric Data object
        graph = Data(x=node_features, edge_index=edge_index, y=labels_tensor)
        graphs.append(graph)

    return graphs

# টেম্পোরাল গ্রাফ তৈরি করুন (শুধু ট্রেইন ডেটার জন্য)
time_steps = 20
print(f"⏳ {time_steps} টাইমস্টেপে গ্রাফ তৈরি হচ্ছে...")
train_graphs = create_temporal_graph(X_train, y_train, time_steps=time_steps)
print(f"✅ {len(train_graphs)} টি গ্রাফ তৈরি হয়েছে")

# 10. TGNN মডেল ডিফাইন করা
# ============================================================
print("\n🧠 TGNN মডেল তৈরি করা হচ্ছে...")

class TemporalGNN(nn.Module):
    def __init__(self, input_dim, hidden_dim=128, output_dim=2, num_layers=2):
        super(TemporalGNN, self).__init__()
        self.num_layers = num_layers

        # GCN লেয়ার
        self.convs = nn.ModuleList()
        self.convs.append(GCNConv(input_dim, hidden_dim))
        for _ in range(num_layers - 1):
            self.convs.append(GCNConv(hidden_dim, hidden_dim))

        # LSTM লেয়ার (টেম্পোরাল প্যাটার্ন শেখার জন্য)
        self.lstm = nn.LSTM(hidden_dim, hidden_dim, batch_first=True, num_layers=2, dropout=0.2)

        # ক্লাসিফিকেশন লেয়ার
        self.fc1 = nn.Linear(hidden_dim, 64)
        self.fc2 = nn.Linear(64, output_dim)
        self.dropout = nn.Dropout(0.3)

    def forward(self, graphs):
        graph_features = []

        for graph in graphs:
            x = graph.x
            edge_index = graph.edge_index

            # GCN লেয়ার
            for i, conv in enumerate(self.convs):
                x = conv(x, edge_index)
                if i < len(self.convs) - 1:
                    x = F.relu(x)
                    x = self.dropout(x)

            # গ্লোবাল পুলিং
            x = global_mean_pool(x, torch.zeros(x.size(0), dtype=torch.long, device=x.device))
            graph_features.append(x.unsqueeze(0))

        # টেম্পোরাল ডেটা হিসেবে স্ট্যাক
        if len(graph_features) == 0:
            return torch.zeros((1, 2), device=x.device)

        temporal_data = torch.cat(graph_features, dim=1)

        # LSTM
        lstm_out, _ = self.lstm(temporal_data)

        # ক্লাসিফিকেশন
        x = F.relu(self.fc1(lstm_out[:, -1, :]))
        x = self.dropout(x)
        output = self.fc2(x)

        return output

# মডেল ইনিশিয়ালাইজ করুন
input_dim = X_train.shape[1]
model = TemporalGNN(input_dim=input_dim, hidden_dim=128, output_dim=2, num_layers=2)

# ডিভাইস সেট করুন (GPU থাকলে GPU ব্যবহার করবে)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)
print(f"✅ মডেল তৈরি হয়েছে: {device}")

# 11. ট্রেইনিং ফাংশন
# ============================================================
print("\n🚀 মডেল ট্রেইনিং শুরু হচ্ছে...")

def train_model(model, graphs, epochs=30, lr=0.001):
    """
    মডেল ট্রেইন করুন
    """
    optimizer = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=1e-5)
    scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=10, gamma=0.5)
    criterion = nn.CrossEntropyLoss()

    # ক্লাস ওয়েট (ইমব্যালেন্সের জন্য)
    # y_train থেকে ক্লাস ওয়েট ক্যালকুলেট করুন
    y_train_flat = np.concatenate([g.y.numpy() for g in graphs])
    unique, counts = np.unique(y_train_flat, return_counts=True)
    class_weights = torch.tensor([counts.sum() / (len(unique) * counts[i]) for i in range(len(unique))], dtype=torch.float).to(device)

    losses = []
    accuracies = []
    f1_scores = []

    model.train()

    for epoch in range(epochs):
        optimizer.zero_grad()

        # ফরওয়ার্ড পাস
        output = model(graphs)

        # লেবেল প্রস্তুত করুন
        all_labels = torch.cat([g.y for g in graphs], dim=0).to(device)

        # আউটপুটের সাথে লেবেল মিলিয়ে নিন
        min_len = min(output.size(0), all_labels.size(0))
        output = output[:min_len]
        labels = all_labels[:min_len]

        # লস ক্যালকুলেশন
        loss = criterion(output, labels)
        loss.backward()
        optimizer.step()
        scheduler.step()

        # অ্যাকুরেসি
        pred = output.argmax(dim=1)
        acc = (pred == labels).float().mean().item()

        # F1 স্কোর
        f1 = f1_score(labels.cpu().numpy(), pred.cpu().numpy(), average='weighted')

        losses.append(loss.item())
        accuracies.append(acc)
        f1_scores.append(f1)

        if (epoch + 1) % 5 == 0:
            print(f'Epoch {epoch+1}/{epochs} | Loss: {loss.item():.4f} | Acc: {acc:.4f} | F1: {f1:.4f}')

    return losses, accuracies, f1_scores

# মডেল ট্রেইন করুন
losses, accuracies, f1_scores = train_model(model, train_graphs, epochs=30)

# 12. ইভালুয়েশন
# ============================================================
print("\n📊 মডেল ইভালুয়েশন করা হচ্ছে...")

def evaluate_model(model, X_test, y_test, scaler, time_steps=10):
    """
    টেস্ট ডেটাতে মডেল ইভালুয়েট করুন
    """
    model.eval()

    # টেস্ট ডেটা স্কেল করুন
    X_test_scaled = scaler.transform(X_test)

    # টেস্ট গ্রাফ তৈরি করুন
    test_graphs = create_temporal_graph(X_test_scaled, y_test, time_steps=time_steps)

    with torch.no_grad():
        # গ্রাফগুলো ডিভাইসে নিয়ে যান
        for graph in test_graphs:
            graph.x = graph.x.to(device)
            graph.edge_index = graph.edge_index.to(device)
            graph.y = graph.y.to(device)

        output = model(test_graphs)
        pred = output.argmax(dim=1)

        # লেবেল
        all_labels = torch.cat([g.y for g in test_graphs], dim=0).to(device)
        min_len = min(output.size(0), all_labels.size(0))
        output = output[:min_len]
        labels = all_labels[:min_len]
        pred = pred[:min_len]

        # মেট্রিক্স
        acc = (pred == labels).float().mean().item()
        f1 = f1_score(labels.cpu().numpy(), pred.cpu().numpy(), average='weighted')

        print(f"\n✅ টেস্ট রেজাল্ট:")
        print(f"   Accuracy: {acc:.4f}")
        print(f"   F1-Score: {f1:.4f}")

        # কনফিউশন ম্যাট্রিক্স
        print("\n📊 কনফিউশন ম্যাট্রিক্স:")
        print(confusion_matrix(labels.cpu().numpy(), pred.cpu().numpy()))

        # ক্লাসিফিকেশন রিপোর্ট
        print("\n📋 ক্লাসিফিকেশন রিপোর্ট:")
        print(classification_report(labels.cpu().numpy(), pred.cpu().numpy(),
                                   target_names=['Benign', 'Attack']))

        return acc, f1

# টেস্ট ডেটা তৈরি করুন
print("🔄 টেস্ট ডেটা প্রস্তুত করা হচ্ছে...")
test_acc, test_f1 = evaluate_model(model, X_test, y_test, scaler, time_steps=10)

# 13. ভিজুয়ালাইজেশন
# ============================================================
print("\n📈 ভিজুয়ালাইজেশন তৈরি করা হচ্ছে...")

plt.figure(figsize=(15, 5))

# Loss প্লট
plt.subplot(1, 3, 1)
plt.plot(losses)
plt.title('Training Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True)

# Accuracy প্লট
plt.subplot(1, 3, 2)
plt.plot(accuracies)
plt.title('Training Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.grid(True)

# F1 Score প্লট
plt.subplot(1, 3, 3)
plt.plot(f1_scores)
plt.title('Training F1 Score')
plt.xlabel('Epoch')
plt.ylabel('F1 Score')
plt.grid(True)

plt.tight_layout()
plt.savefig('training_metrics.png', dpi=150)
plt.show()
print("✅ ভিজুয়ালাইজেশন সেভ করা হয়েছে: training_metrics.png")

# 14. মডেল সেভ করা
# ============================================================
print("\n💾 মডেল সেভ করা হচ্ছে...")

# মডেল সেভ
torch.save({
    'model_state_dict': model.state_dict(),
    'scaler': scaler,
    'feature_cols': feature_cols,
    'input_dim': input_dim
}, 'temporal_gnn_model_1core.pth')
print("✅ মডেল সেভ হয়েছে: temporal_gnn_model_1core.pth")

# Google Drive-এ সেভ করতে চাইলে (ঐচ্ছিক)
from google.colab import drive
drive.mount('/content/drive')
!cp temporal_gnn_model_1core.pth /content/drive/MyDrive/
print("✅ মডেল Google Drive-এ সেভ হয়েছে!")

# 15. ফাইনাল রিপোর্ট
# ============================================================
print("\n" + "="*60)
print("🎉 ট্রেইনিং সম্পূর্ণ!")
print("="*60)
print(f"✅ মোট ডেটা: {TOTAL_SAMPLES:,}")
print(f"✅ ট্রেইন ডেটা: {X_train.shape[0]:,}")
print(f"✅ টেস্ট ডেটা: {X_test.shape[0]:,}")
print(f"✅ ফিচার সংখ্যা: {X_train.shape[1]}")
print(f"✅ টেস্ট Accuracy: {test_acc:.4f}")
print(f"✅ টেস্ট F1-Score: {test_f1:.4f}")
print("="*60)
```

```python
"""
CIC-IoT-2023 ডেটাসেট লোকাল পিসিতে ডাউনলোড করার স্ক্রিপ্ট
"""

import os
import sys
import time
from datasets import load_dataset
import shutil

def download_dataset():
    """
    ডেটাসেট ডাউনলোড করার মূল ফাংশন
    """
    print("=" * 70)
    print("🚀 CIC-IoT-2023 ডেটাসেট ডাউনলোডার (লোকাল পিসি)")
    print("=" * 70)

    # ডাউনলোডের সময় দেখানোর জন্য
    start_time = time.time()

    try:
        # কনফিগারেশন নির্বাচন করুন
        config_name = "random_3way"  # random_3way ভালো (ট্রেইন+টেস্ট+ভ্যালিডেশন)

        print(f"\n📥 কনফিগারেশন: {config_name}")
        print("📥 ডেটাসেট লোড হচ্ছে... (এতে ১০-৩০ মিনিট সময় লাগতে পারে)")
        print("📥 মোট ডাউনলোড সাইজ: ~১.৬৩ GB")
        print("📥 ডাউনলোড লোকেশন: আপনার পিসির ক্যাশ ফোল্ডার\n")

        # ডেটাসেট ডাউনলোড করুন
        dataset = load_dataset(
            "lacg030175/CIC-IoT-2023-neto-full",
            config_name
        )

        # ডাউনলোডের সময় গণনা
        elapsed_time = time.time() - start_time
        minutes = int(elapsed_time // 60)
        seconds = int(elapsed_time % 60)

        print("\n" + "=" * 70)
        print("✅ ডাউনলোড সম্পূর্ণ!")
        print(f"⏱️  সময় লেগেছে: {minutes} মিনিট {seconds} সেকেন্ড")
        print("=" * 70)

        # ডেটাসেটের তথ্য দেখান
        print("\n📊 ডেটাসেট তথ্য:")
        print("-" * 50)
        for split_name, split_data in dataset.items():
            print(f"  {split_name}: {split_data.num_rows:,} সারি")
            print(f"    ফিচার: {len(split_data.features)}টি")

        # ফিচারের তালিকা দেখান
        print("\n📋 ফিচারের তালিকা (প্রথম ১০টি):")
        print("-" * 50)
        features = list(dataset['train'].features.keys())
        for i, feature in enumerate(features[:10], 1):
            print(f"  {i:2d}. {feature}")
        print(f"  ... এবং আরও {len(features) - 10}টি ফিচার")

        # নমুনা ডেটা দেখান
        print("\n🔍 নমুনা ডেটা (প্রথম ৩টি সারি):")
        print("-" * 50)
        sample = dataset['train'].select(range(3))
        for i, row in enumerate(sample):
            print(f"\n  সারি {i+1}:")
            print(f"    attack_class: {row['attack_class']}")
            print(f"    label: {row['label']} (0=Benign, 1=Attack)")
            print(f"    Protocol Type: {row['Protocol Type']}")
            print(f"    TCP: {row['TCP']}, UDP: {row['UDP']}")

        # ডেটাসেটের অবস্থান দেখান
        print("\n📁 ডেটাসেট সেভ হয়েছে:")
        print("-" * 50)

        # ক্যাশ ডিরেক্টরি লোকেশন
        cache_dir = os.path.expanduser("~/.cache/huggingface/datasets")
        print(f"  📂 {cache_dir}")

        # ফাইলের সাইজ দেখান
        total_size = 0
        if os.path.exists(cache_dir):
            for root, dirs, files in os.walk(cache_dir):
                for f in files:
                    fp = os.path.join(root, f)
                    if os.path.exists(fp):
                        total_size += os.path.getsize(fp)

        print(f"  📊 মোট ডাউনলোড সাইজ: {total_size / (1024**3):.2f} GB")
        print("\n  ⚠️  পরবর্তীতে এই ডেটাসেট অফলাইনেও কাজ করবে!")
        print("  💡 ডেটাসেট আপনার পিসির ক্যাশে সংরক্ষিত আছে")

        return dataset

    except Exception as e:
        print("\n❌ ডাউনলোডে সমস্যা হয়েছে!")
        print(f"ত্রুটি: {e}")
        print("\nসম্ভাব্য সমাধান:")
        print("1. ইন্টারনেট কানেকশন চেক করুন")
        print("2. পর্যাপ্ত স্টোরেজ আছে কিনা দেখুন (প্রয়োজন ~৩.২৬ GB)")
        print("3. নিচের কমান্ড দিয়ে লাইব্রেরি আপডেট করুন:")
        print("   pip install --upgrade datasets huggingface_hub")
        return None

def check_storage():
    """
    পর্যাপ্ত স্টোরেজ আছে কিনা চেক করে
    """
    try:
        import shutil
        total, used, free = shutil.disk_usage("/")
        free_gb = free // (2**30)

        print("\n💾 স্টোরেজ তথ্য:")
        print("-" * 50)
        print(f"  ফ্রি স্পেস: {free_gb} GB")

        if free_gb < 5:
            print("  ⚠️  সতর্কতা: ফ্রি স্পেস কম! ডাউনলোডের জন্য ন্যূনতম ৫ GB প্রয়োজন।")
            return False
        else:
            print("  ✅ পর্যাপ্ত স্পেস আছে।")
            return True
    except:
        print("  ⚠️  স্টোরেজ চেক করা সম্ভব হয়নি")
        return True

def main():
    """
    মেইন ফাংশন
    """
    print("\n" + "=" * 70)
    print("🚀 CIC-IoT-2023 ডেটাসেট ডাউনলোডার (লোকাল পিসি)")
    print("=" * 70)

    # স্টোরেজ চেক
    if not check_storage():
        proceed = input("\nতবুও ডাউনলোড চালাবেন? (y/n): ")
        if proceed.lower() != 'y':
            print("ডাউনলোড বাতিল করা হয়েছে।")
            return

    # টোকেন চেক (ঐচ্ছিক)
    print("\n🔑 Hugging Face টোকেন (ঐচ্ছিক):")
    print("  (টোকেন দিলে ডাউনলোডের গতি বাড়ে)")
    print("  টোকেন পেতে: https://huggingface.co/settings/tokens")

    token = input("\nটোকেন থাকলে দিন (না থাকলে Enter চাপুন): ").strip()

    if token:
        try:
            from huggingface_hub import login
            login(token=token)
            print("✅ টোকেন সেট করা হয়েছে!")
        except:
            print("⚠️  টোকেন সেট করতে সমস্যা হয়েছে, টোকেন ছাড়াই চালানো হচ্ছে...")

    # ডাউনলোড শুরু
    dataset = download_dataset()

    if dataset:
        print("\n" + "=" * 70)
        print("🎉 সবকিছু সম্পূর্ণ! এখন আপনি ডেটাসেট ব্যবহার করতে পারেন।")
        print("=" * 70)

        # ডেটাসেট ব্যবহারের উদাহরণ
        print("\n📝 ডেটাসেট ব্যবহারের উদাহরণ (আপনার পিসিতে):")
        print("-" * 50)
        print("""
# ডেটাসেট লোড করুন (ইতিমধ্যে ডাউনলোড হয়ে থাকলে দ্রুত লোড হবে)
from datasets import load_dataset
dataset = load_dataset("lacg030175/CIC-IoT-2023-neto-full", "random_3way")

# ট্রেইন ডেটা থেকে প্যান্ডাস ডেটাফ্রেম তৈরি করুন
df = dataset['train'].select(range(1000)).to_pandas()
print(df.head())

# ডেটাসেটের ক্যাশ লোকেশন জানতে
import os
print(os.path.expanduser("~/.cache/huggingface/datasets"))
        """)

        print("\n💡 ডেটাসেট কোথায় আছে মনে রাখবেন:")
        print(f"   {os.path.expanduser('~/.cache/huggingface/datasets')}")

if __name__ == "__main__":
    main()
```
